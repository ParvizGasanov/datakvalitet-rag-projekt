```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/original_data.csv")

df
```

```python
cleaned_df = df.copy()

cleaned_df
```

```python
cleaned_df.columns = (
    cleaned_df.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)

cleaned_df.columns
```

```python
cleaned_df = cleaned_df.drop_duplicates()

cleaned_df
```

```python
def row_to_text(row):
    return f"""
    Titel: {row['title']}
    Typ: {row['type']}
    Land: {row['country']}
    År: {row['release_year']}
    Genre: {row['genre']}
    Beskrivning: {row['description']}
    """

cleaned_df["text"] = cleaned_df.apply(row_to_text, axis=1)

cleaned_df[["title", "text"]]
```

```python
cleaned_df.to_csv("../data/cleaned_data.csv", index=False)
```
